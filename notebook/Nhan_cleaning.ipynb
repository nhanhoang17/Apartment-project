EDA for apartment dataset:

```python
import pandas as pd

df = pd.read_csv(
    'apartments_for_rent_classified_10K.csv',
    sep=';',
    encoding='cp1252',
    keep_default_na=False,   
    na_values=['', 'null']   
)
pd.set_option('display.max_columns', None)
df.head()
```

```python
pets_nan_row=df['pets_allowed'].isna().sum()

total=len(df)

rate_pets=pets_nan_row/total
```

```python
# tỉ lệ dòng bị NaN của pets trên tổng số.
rate_pets*100
```

```python
amenities_nan_row=df['amenities'].isna().sum()

rate_amenities=amenities_nan_row/total 
```

```python
# tỉ lệ dòng bị NaN của amenities trên tổng số.
rate_amenities
```
