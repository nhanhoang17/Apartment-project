```python
import pandas as pd
pd.set_option('display.max_rows', None)

df = pd.read_csv('data/apartments_for_rent_classified_10K.csv', sep=';', encoding='cp1252')
```

```python
len(df.columns.to_list())
```

```python
df.head(10)
```

```python
df.drop(columns='fee', inplace=True)
```

```python
amenities = set()

for value in df['amenities']:
    if pd.notna(value):
        amenities.update(value.split(','))

print(amenities)
print(len(amenities))
```

```python
df['num_amenities'] = df['amenities'].apply(
    lambda x: len(x.split(',')) if pd.notna(x) else 0
)

print(df['num_amenities'].head(10))

df['has_amenities'] = (df['num_amenities'] > 0).astype(int)

print(df['has_amenities'].head(10))
```

```python
df['has_photo'] = df['has_photo'].apply(
    lambda x: 0 if x == 'No' else 1 if x == 'Thumbnail' else 2
)

df['has_photo'].head(10)
```

```python
df['num_pets_allowed'] = df['pets_allowed'].apply(
    lambda x: len(x) if pd.notna(x) else 0
)
```
